```scala
%spark.pyspark
from pyspark.sql import SparkSession
from pyspark.sql.functions import explode, col, row_number, size, to_date, to_timestamp, round
from pyspark.sql.window import Window
from pyspark.sql.types import DateType, IntegerType
from pyspark.sql import functions as F
spark = SparkSession.builder \
    .appName("Transformations") \
    .enableHiveSupport() \
    .getOrCreate()
```

```scala
%spark.pyspark
df = spark.read.option("multiLine", True).json("/user/root/raw_orders.json")
```

```scala
%spark.pyspark
before_count = df.select(size(col("items")).alias("num_items")) \
                 .agg({"num_items": "sum"}).collect()[0][0]
print("Total items BEFORE explode:", before_count)
```

```scala
%spark.pyspark
flattened_df = df.withColumn("item", explode(col("items"))) \
    .select(
        col("order_id"),
        col("customer_id"),
        col("order_date"),
        col("total_value"),
        col("payment_status"),
        col("item.product_id").alias("product_id"),
        col("item.sales_quantity").alias("sales_quantity")
    )
```

```scala
%spark.pyspark
flattened_df = flattened_df.withColumn("order_date",to_date(to_timestamp(col("order_date"), "M/d/yyyy'T'H:mm'Z'"))
).withColumn("total_value", round(col("total_value"), 2)
).withColumn("sales_quantity", col("sales_quantity").cast(IntegerType()))
```

```scala
%spark.pyspark
window_spec = Window.partitionBy("order_id").orderBy("product_id")
flattened_df = flattened_df.withColumn("order_line_id", row_number().over(window_spec))
```

```scala
%spark.pyspark
null_products = flattened_df.filter(col("product_id").isNull()).count()
print("Null product_id rows:", null_products)

# - Check for invalid sales_quantity
invalid_qty = flattened_df.filter(col("sales_quantity") <= 0).count()
print("Invalid sales_quantity rows:", invalid_qty)

# - Check consistency of counts after explode
after_count = flattened_df.count()
print("Total items AFTER explode:", after_count)
print("Check Match:", before_count == after_count)
```

```scala
%spark.pyspark
spark.sql("CREATE DATABASE IF NOT EXISTS orders")
spark.catalog.setCurrentDatabase("orders")
```

```scala
%spark.pyspark
flattened_df.write.format("parquet") \
    .option("path", "hdfs://namenode:9000/user/hive/warehouse/orders.db/staging_orderlines") \
    .option("compression", "snappy") \
    .mode("overwrite") \
    .saveAsTable("orders.staging_orderlines")
```

```scala
%spark.pyspark
df = flattened_df
max_date = df.agg(F.max("order_date")).collect()[0][0]

rfm_df = df.groupBy("customer_id").agg(
    F.max("order_date").alias("last_purchase"),
    F.countDistinct("order_id").alias("frequency"),
    F.sum("total_value").alias("monetary")
)

rfm_df = rfm_df.withColumn("recency", F.datediff(F.lit(max_date), F.col("last_purchase")))

rfm_df = rfm_df.select("customer_id", "recency", "frequency", "monetary")

rfm_df.show(20, truncate=False)
```

```scala
%spark.pyspark
rfm_df.write.format("parquet") \
    .option("path", "hdfs://namenode:9000/user/hive/warehouse/orders.db/RFM") \
    .option("compression", "snappy") \
    .mode("overwrite") \
    .saveAsTable("orders.RFM")
```

```scala
%spark.pyspark
dim_date = (
    spark.table("staging_orderlines")  
    .select("order_date")              
    .distinct()                        
    .withColumn("date_key", F.col("order_date"))
    .withColumn("year", F.year("order_date"))
    .withColumn("month", F.month("order_date"))
    .withColumn("day", F.dayofmonth("order_date"))
    .withColumn("quarter", F.quarter("order_date"))
    .withColumn("day_of_week", F.dayofweek("order_date"))
    .withColumn("week_of_year", F.weekofyear("order_date"))
    .withColumn(
        "day_type",
        F.when(F.dayofweek("order_date").isin([1, 7]), "Weekend").otherwise("Weekday")
    )
)
```

```scala
%spark.sql
SELECT
    DATE_FORMAT(order_date, "yyyy-MM") AS year_month,
    COUNT(DISTINCT order_id) AS total_orders,
    SUM(total_value) AS total_revenue
FROM staging_orderlines
GROUP BY DATE_FORMAT(order_date, "yyyy-MM")
ORDER BY year_month
```

```scala
%spark.sql
SELECT 
    date_format(order_date, 'E') AS day_of_week,
    COUNT(DISTINCT order_id) AS total_orders,
    SUM(total_value) AS total_revenue
FROM staging_orderlines
GROUP BY date_format(order_date, 'E')
ORDER BY total_orders DESC
```
