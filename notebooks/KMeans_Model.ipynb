```scala
%spark.pyspark

from pyspark.ml.feature import VectorAssembler, StandardScaler
from pyspark.ml.clustering import KMeans
from pyspark.sql.functions import col
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
from mpl_toolkits.mplot3d import Axes3D
```

```scala
%spark.pyspark
spark = SparkSession.builder \
    .appName("KMeand Model") \
    .enableHiveSupport() \
    .getOrCreate()
```

```scala
%spark.pyspark

rfm_df = spark.sql("SELECT * FROM orders.rfm")

# Assemble Features

assembler = VectorAssembler(
    inputCols=["recency", "frequency", "monetary"], 
    outputCol="features"
)
assembled = assembler.transform(rfm_df)

# Scale Features

scaler = StandardScaler(
    inputCol="features", 
    outputCol="scaled_features", 
    withStd=True, 
    withMean=True
)
scaler_model = scaler.fit(assembled)
scaled_data = scaler_model.transform(assembled)


# Elbow Method
cost = []
K = range(2, 10)

print("Running Elbow Method...")
for k in K:
    kmeans = KMeans(featuresCol="scaled_features", k=k, seed=20)
    model = kmeans.fit(scaled_data)
    cost.append(model.summary.trainingCost)
    print(f"k={k}, Cost={model.summary.trainingCost:.2f}")


# Plot Elbow
plt.figure(figsize=(10, 6))
plt.plot(K, cost, 'bx-')
plt.xlabel('k (Number of clusters)')
plt.ylabel('Cost (WSSSE)')
plt.title('Elbow Method For Optimal k')
plt.grid(True)
plt.show()
```

```scala
%spark.pyspark

# Train KMeans with optimal k

optimal_k = 4 

kmeans = KMeans(
    featuresCol="scaled_features", 
    predictionCol="cluster", 
    k=optimal_k, 
    seed=20
)
model = kmeans.fit(scaled_data)

clustered = model.transform(scaled_data)

print(f"✓ KMeans trained with k={optimal_k}")
print("\nSample of clustered customers:")
clustered.select("customer_id", "recency", "frequency", "monetary", "cluster").show(20)

# Cluster Centers

print("\nCluster Centers:")
for idx, center in enumerate(model.clusterCenters()):
    print(f"Cluster {idx}: Recency={center[0]:.2f}, Frequency={center[1]:.2f}, Monetary={center[2]:.2f}")

centers_df = pd.DataFrame(
    model.clusterCenters(), 
    columns=["recency", "frequency", "monetary"]
)
print("\nCenters DataFrame:")
print(centers_df)
```

```scala
%spark.pyspark

# Convert to Pandas for plotting

pdf = clustered.select("customer_id", "recency", "frequency", "monetary", "cluster").toPandas()

cluster_names = {
    0: "Regulars",
    1: "Loyal",
    2: "Big Spenders / VIP",
    3: "Churned / Inactive"
}

pdf["cluster_name"] = pdf["cluster"].map(cluster_names)

print(f"✓ Data converted to Pandas ({len(pdf)} rows)")
print("\nCluster distribution:")
print(pdf["cluster_name"].value_counts())
```

```scala
%spark.pyspark

# Visualization: Frequency vs Monetary

plt.figure(figsize=(10, 7))
for cname in pdf["cluster_name"].unique():
    cluster_points = pdf[pdf["cluster_name"] == cname]
    plt.scatter(
        cluster_points["frequency"], 
        cluster_points["monetary"],
        label=cname, 
        alpha=0.6, 
        s=50
    )

plt.xlabel("Frequency", fontsize=12)
plt.ylabel("Monetary", fontsize=12)
plt.title("Customer Segmentation (Frequency vs Monetary)", fontsize=14)
plt.legend(title="Cluster")
plt.gca().yaxis.set_major_formatter(mtick.StrMethodFormatter('{x:,.0f}'))
plt.grid(True, alpha=0.3)
plt.show()
```
